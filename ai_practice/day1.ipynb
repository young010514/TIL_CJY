# itertools 
- 반복 관련 작업을 간단하게 처리해주는 라이브러리

```python
from itertools import product

for i in product([1,2],["A","B"]):
    print(i)
    # 데카르트 곱 (모든 값들을 교차해서 생성)
```

```python
for i in product([1,2,3],repeat=2):
    print(i)
    # repeat 반복횟수 : 반복횟수만큼 중복순열
```

```python
from itertools import product, permutations, combinations, accumulate # 중복순열, 순열, 조합, 누적합
```

```python
for p in permutations([1,2,3],2):# 순열 ( 금은동)
    print(p)
```

```python
for c in combinations([1,2,3],2):
    print(c)
```

```python
nums = [1,2,3,4]

print("누적합:",list(accumulate(nums)))
```

```python
import operator   # 누적곱 구할때 필요한 module
nums = [1,2,3,4]

print("누적곱:",list(accumulate(nums,func=operator.mul)))
```

### 2차원 행렬을 다시 1차원 벡터로 만들기
1. 방법 1 (얕은 복사 또는 깊은 복사 활용)
    - ravel() : 얕은 복사
    - flatten() : 깊은 복사
2. 방법 2 (
   - reshape()




```python
# 방법1
vect = np.array([1,2,3])

vect = np.tile(vect,(3,1))
arr = vect.ravel()
print(arr)
# 얕은 복사로 인해 값이 변동됨
vect[0][0] = 4
print(arr)


arr = vect.flatten()
print(arr)
vect[0][0] = 5
print(arr)
```

```python
# 방법2
vect = np.array(range(30))
vect = vect.reshape(-1)  # -1이 의미하는 것은 자동계산
print(vect)
arr = vect.reshape(3,-1)   # 행이 3개, 그 뒤론 자동 계산
print(arr)
arr = vect.reshape(3,5,2)
print(arr)
```

### loc vs iloc
- loc = 라벨 기반의 인덱스
- iloc = 위치 기반의 인덱스

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "Alice", "Bob"],
    "Subject": ["Math", "Math", "Math", "English", "English"],
    "Score": [85, 90, 95, 80, 88]
})

print(df)
```

```python
# 조건 필터링
# 영어 과목 정보만 출력해보기
print(df[df["Subject"] == "English"])
```

```python
print(df[(df["Subject"] == "Math") & (df["Score"] >= 90)])
# and 나 or 연산자 넣을려면 조건문 마다 소괄호 추가
```

### groupby


```python
# Subject 열의 값이 같은 것 끼리 묶고
# 각 Subject 별로 평균 점수를 계산

group = df.groupby("Subject")["Score"].mean()
print(group)
```

```python
group.name = "Mean"
print(group)
```
